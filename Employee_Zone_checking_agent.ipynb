```python
from google.colab import files

uploaded = files.upload()
```

```python
import pandas as pd

file_name = list(uploaded.keys())[0]

df = pd.read_csv(file_name)

df.head()
```

```python
from datetime import datetime

def classify_employee(time_str):

    check_time = datetime.strptime(
        time_str,
        "%H:%M"
    )

    cutoff = datetime.strptime(
        "09:00",
        "%H:%M"
    )

    if check_time < cutoff:
        return "SAFE ZONE"

    return "DANGER ZONE"
```

```python
df_long = df.melt(id_vars=['Date'], var_name='Employee', value_name='CheckIn_InOut')

df_long['CheckIn'] = df_long['CheckIn_InOut'].apply(lambda x: x.split('-')[0] if pd.notna(x) else None)

df_long['Status'] = df_long['CheckIn'].apply(classify_employee)

df = df_long

df.head()
```

```python
with open(
    "employee_report.txt",
    "w",
    encoding="utf-8"
) as f:

    f.write("EMPLOYEE ATTENDANCE REPORT\n")
    f.write("="*50 + "\n\n")

    for _, row in df.iterrows():

        f.write(
            f"Employee: {row['Employee']}\n"
        )

        f.write(
            f"Check-In: {row['CheckIn']}\n"
        )

        f.write(
            f"Status: {row['Status']}\n"
        )

        f.write("-"*40 + "\n")
```

```python
# Save the DataFrame to a new CSV file
df.to_csv('employee_status_report.csv', index=False)
print("DataFrame saved to 'employee_status_report.csv'")
```
